```python
import pandas as pd
import numpy as np
```

```python
# --- Fill in paths to your CSVs and labels ---
csv_paths = [
    "/path/to/model_a_metrics.csv",
    "/path/to/model_b_metrics.csv",
]
labels = [
    "Model A",
    "Model B",
]
```

```python
# Expected columns: tomo_name, precision, recall, dice
metrics = ["precision", "recall", "dice"]

for label, path in zip(labels, csv_paths):
    df = pd.read_csv(path)

    mean_row = df[metrics].mean().to_frame().T
    mean_row.insert(0, "tomo_name", "Mean")

    table = pd.concat([df[["tomo_name"] + metrics], mean_row], ignore_index=True)
    table[metrics] = table[metrics].round(3)
    table = table.set_index("tomo_name")
    table.index.name = None
    table.columns = [c.capitalize() for c in table.columns]

    mean_idx = table.index.get_loc("Mean")

    styled = (
        table.style
        .set_caption(label)
        .set_table_styles([
            # outer border top and bottom
            {"selector": "table",    "props": "border-collapse: collapse; font-family: sans-serif; font-size: 13px;"},
            {"selector": "caption",  "props": "font-size: 14px; font-weight: bold; text-align: left; padding-bottom: 6px;"},
            {"selector": "thead th", "props": "border-top: 2px solid black; border-bottom: 1px solid black; padding: 6px 14px; text-align: center;"},
            {"selector": "td",       "props": "padding: 5px 14px; text-align: center; border: none;"},
            {"selector": "th",       "props": "text-align: left; padding: 5px 14px;"},
            {"selector": "tbody tr:last-child td", "props": "border-top: 1px solid black; border-bottom: 2px solid black; font-weight: bold;"},
            {"selector": "tbody tr:last-child th", "props": "border-top: 1px solid black; border-bottom: 2px solid black; font-weight: bold;"},
        ])
        .format("{:.3f}")
    )

    display(styled)
```
